```python
# Informações coluna de Número de filhos do cliente.

df = pd.DataFrame({'numero_filhos': [0, 1, 2, 0, 1, 3, 0, 2, 4, 1, 0, 2]})

# Coleta dos principais parâmetros .
resumo = df['numero_filhos'].describe()
print(resumo)

# Funções isoladas úteis
#moda = df['numero_filhos'].mode()[0]
#amplitude = df['numero_filhos'].max() - df['numero_filhos'].min()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calcular as proporções para a coluna de exemplo
prop_exemplo = df['numero_filhos'].value_counts(normalize=True).rename('proporcao').reset_index()
prop_exemplo['Origem'] = 'Exemplo (numero_filhos)'
prop_exemplo.rename(columns={'numero_filhos': 'Qtd_Filhos'}, inplace=True)

# Calcular as proporções para a coluna real
prop_real = df_todos_nan['CL_FHL'].value_counts(normalize=True).rename('proporcao').reset_index()
prop_real['Origem'] = 'Real (CL_FHL)'
prop_real.rename(columns={'CL_FHL': 'Qtd_Filhos'}, inplace=True)

# Unificar os dados para o gráfico
df_comparativo = pd.concat([prop_exemplo, prop_real])

# Plotar o gráfico de barras comparativo
plt.figure(figsize=(10, 6))
sns.barplot(data=df_comparativo, x='Qtd_Filhos', y='proporcao', hue='Origem', palette='Set2')

# Customizações do gráfico
plt.title('Comparação da Proporção do Número de Filhos', fontsize=14)
plt.xlabel('Quantidade de Filhos', fontsize=12)
plt.ylabel('Proporção (Frequência Relativa)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Base de Dados')

plt.show()
```
